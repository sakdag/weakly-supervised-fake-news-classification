```python
from sklearn.model_selection import train_test_split
```

```python
df.sample(frac=1).reset_index(drop = True)
df_train, df_test = train_test_split(df, train_size=0.8)
```

```python
from snorkel.labeling import labeling_function, PandasLFApplier, LFAnalysis
```

```python
import re
```

```python
QUESTION = 1
QUOTE = 0
ABSTAIN = -1
```

```python
@labeling_function()
def lf_keyword_lookup(x):
  keywords = ["why", "when", "who", "what", "how", "where"]
  return QUESTION if any(word in x.sentences.lower() for word in keywords) else ABSTAIN
```

```python
@labeling_function()
def lf_regex_contains_what(x):
  return QUESTION if re.search(r"what*?", x.sentences, flags=re.I) else ABSTAIN
```

```python
lfs = [lf_keyword_lookup, lf_regex_contains_what]
applier = PandasLFApplier(lfs)
L_train = applier.apply(df_train)
```

```python
L_train
```

```python
LFAnalysis(L=L_train, lfs=lfs).lf_summary()
```

```python
from snorkel.labeling.model import MajorityLabelVoter, LabelModel
```

```python
label_model = LabelModel(cardinality=2, verbose=True)
label_model.fit(L_train=L_train, n_epochs=500, log_freq=100, seed=1234)
```
